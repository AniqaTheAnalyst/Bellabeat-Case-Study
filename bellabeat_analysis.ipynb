```python
print("Needed Libraries Loaded")

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
print("Loaded all the needed datasets")

# Folder 1 (March - April)
daily_activity_1 = pd.read_csv('/kaggle/input/datasets/aniqaraihana/bellabeat-casestudy-fitbit-data1/dailyActivity_merged.csv')
hourly_steps_1 = pd.read_csv('/kaggle/input/datasets/aniqaraihana/bellabeat-casestudy-fitbit-data1/hourlySteps_merged.csv')
weigth_log_info_1 = pd.read_csv('/kaggle/input/datasets/aniqaraihana/bellabeat-casestudy-fitbit-data1/weightLogInfo_merged.csv')

# Folder 2 (April - May)
daily_activity_2 = pd.read_csv('/kaggle/input/datasets/aniqaraihana/bellabeat-casestudy-fitbit-data1/dailyActivity_merged (2).csv')
hourly_steps_2 = pd.read_csv('/kaggle/input/datasets/aniqaraihana/bellabeat-casestudy-fitbit-data1/hourlySteps_merged (2).csv')
weigth_log_info_2 = pd.read_csv('/kaggle/input/datasets/aniqaraihana/bellabeat-casestudy-fitbit-data1/weightLogInfo_merged (2).csv')
#sleep_Day 
sleep_Day = pd.read_csv('/kaggle/input/datasets/aniqaraihana/bellabeat-casestudy-fitbit-data1/sleepDay_merged(2).csv')
```

```python
# merging TWO folders ( March - April and April - May) - March to May
daily_activity = pd.concat([daily_activity_1 , daily_activity_2 ], ignore_index=True)
hourly_steps = pd.concat([hourly_steps_1 , hourly_steps_2] , ignore_index=True)
weight_log_info = pd.concat([weigth_log_info_1 , weigth_log_info_2] , ignore_index=True)
```

## Removing Duplicates

```python
daily_activity = daily_activity.drop_duplicates()
hourly_steps = hourly_steps.drop_duplicates()
weight_log_info = weight_log_info.drop_duplicates()
sleep_Day = sleep_Day.drop_duplicates()
```

```python
print("=== AFTER REMOVING DUPLICATES ===")
print(f'daily_activity data : {len(daily_activity)}')
print(f'hourly_steps data : {len(hourly_steps)}')
print(f'weight_log_info data : {len(weight_log_info)}')
print(f'sleep_Day data : {len(sleep_Day)}')
```

## Date formats fixed!

```python
daily_activity['ActivityDate'] = pd.to_datetime(daily_activity['ActivityDate'])
```

```python
hourly_steps['ActivityHour'] = pd.to_datetime(hourly_steps['ActivityHour'])
```

```python
weight_log_info['Date'] = pd.to_datetime(weight_log_info['Date'])
```

```python
sleep_Day['SleepDay'] = pd.to_datetime(sleep_Day['SleepDay'])
```

### Removing unnecessary columns

```python
# Drop unnecessary columns from daily_activity
columns_to_drop = [
    'TotalDistance',
    'TrackerDistance', 
    'LoggedActivitiesDistance',
    'VeryActiveDistance',
    'ModeratelyActiveDistance',
    'LightActiveDistance',
    'SedentaryActiveDistance'
]

daily_activity = daily_activity.drop(columns=columns_to_drop)
```

```python
# Drop unnecessary columns from weight_log_info
weight_log_info = weight_log_info.drop(columns=[
    'Fat',
    'WeightPounds', 
    'IsManualReport',
    'LogId'
])
```

```python
# Also drop from sleep
sleep_Day = sleep_Day.drop(columns=['TotalSleepRecords'])
```

```python
daily_activity.to_csv('daily_activity_clean.csv', index=False)
```
